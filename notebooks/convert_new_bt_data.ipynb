```python
def create_params(underlying):
    params = {
        'UNDERLYING':f"{underlying}",
        'file_path':f"/home/cloudcraftz/Music/OneDrive_2_22-04-2025/{underlying}/2024",
        'new_path':f"/home/cloudcraftz/Music/OneDrive_2_22-04-2025/{underlying}/2024_new"
    }

    return params
```

```python
import os
import glob
# from modules.global_variables import params
import pandas as pd

def create_new_bt(params):
    # Create necessary directories for processed data, if they don't exist
    # file_path = os.path.join('Processed_Data',params['UNDERLYING'],params['EXPIRY_TYPE'].split('_')[1].upper(),params['START_DATE_TIME'].split('-')[0])
    # new_path = os.path.join('Processed_Data',params['UNDERLYING'],params['EXPIRY_TYPE'].split('_')[1].upper(),params['START_DATE_TIME'].split('-')[0]+"_backtest_run")

    file_path = params['file_path']
    new_path = params['new_path']

    os.makedirs(new_path, exist_ok=True)
    
    spot_data_path = os.path.join(new_path, "Spot")
    os.makedirs(spot_data_path, exist_ok=True)
    
    file_list = glob.glob(file_path+"/*.csv")
    
    end = "_Intraday_Spot.csv"
    
    for file in file_list:
        try:
            df = pd.read_csv(file)
            pathlist = file.split('/')
            tradingdate = pathlist[-1].split('_')[1]
    
            if params['UNDERLYING'] == 'SPXW':
                spot_df = df[['Date Time', 'Spot', 'Spot_Missing', 'Prev_Spot']]
            else:
                spot_df = df[['Date Time', 'Spot']]
                
            final_spot_df = spot_df.drop_duplicates().reset_index(drop=True)
            spot_file = params['UNDERLYING']+'_'+tradingdate.replace('-', '')+end
            final_spot_df.to_csv(os.path.join(spot_data_path, spot_file), index=False)
    
            for expdate in df['ExpiryDate'].unique():
                df_sep = df[df['ExpiryDate'] == expdate]
                exp_date = expdate.replace("-","")
    
                new_file_name = f"{params['UNDERLYING']}_{tradingdate}_{exp_date}_Intraday_Preprocessed.csv"
                new_path_1 = os.path.join(new_path, new_file_name)
    
                df_sep.to_csv(new_path_1, index=False)
    
        except Exception as e:
            print(file)
```
