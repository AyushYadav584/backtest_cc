```python
import pandas as pd
import plotly.graph_objects as go
from datetime import datetime, time
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("/home/cloudcraftz/Music/2024/ASIANPAINT/2024_new/ASIANPAINT_20240627_20240627_Intraday_Preprocessed.csv")
df.set_index('Date Time', inplace=True)
df.index = pd.to_datetime(df.index)
```

```python
def price_plot(op_strike, op_type):
    ins_df = df[(df['Strike'] == op_strike) & (df['Type'] == op_type)]
    ins_df['mid_price'] = (ins_df['BidPrice'] + ins_df['AskPrice'])/2

    action_C_value = go.Scatter(x=ins_df.index, y=ins_df['mid_price'], mode='lines', name='mid')
    action_B_value = go.Scatter(x=ins_df.index, y=ins_df['BidPrice'], mode='lines', name='BidPrice')
    action_A_value = go.Scatter(x=ins_df.index, y=ins_df['AskPrice'], mode='lines', name='AskPrice')

    fig = go.Figure()
    # fig = make_subplots(specs = [[{"secondary_y": True}]])
    fig.add_trace(action_C_value)
    fig.add_trace(action_A_value)
    fig.add_trace(action_B_value)

    fig.update_layout(
        xaxis = dict(
                        title='Timestamp',
                    ),
            yaxis = dict(
                        title = f'Price',
                    ),

            plot_bgcolor = "white",
            title=f'Price movement of the instrument Strike {op_strike} Option type {op_type} in ASIANPAINT data date 2024-06-27',
        )

    fig.update_xaxes(
                    mirror=True,
                    ticks='outside',
                    showline=True,
                    linecolor='black',
                    gridcolor='white')

    fig.update_yaxes(
                mirror=True,
                ticks='outside',
                showline=True,
                linecolor='black',
                gridcolor='white')

    fig.show()
```

```python
op_type = "CE"
op_strike = 2920

price_plot(op_strike, op_type)
```
