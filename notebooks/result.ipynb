```python
import pandas as pd
from glob import glob
```

```python
fast = 25
slow = 45
```

```python
file_list = glob(f"/home/cloudcraftz/Office_Projects/10_Fintech/backtest_2.0_dev/HFT-Options-EIS-Global/tradelib/outputs/NF_SMA_CROSS_OVER_MFT/2024_NIFTY_SMA_STRADDLE_STRATEGY_fast_{fast}_slow_{slow}_tol_0/consolidated_store/*.csv")
```

```python
Alpha_bt_df_list = [pd.read_csv(i) for i in file_list]

Alpha_df = pd.concat(Alpha_bt_df_list, ignore_index=True)
Alpha_df = Alpha_df[Alpha_df['trade_done']==True]
Alpha_df['Timestamp'] = pd.to_datetime(Alpha_df['timestamp'])
Alpha_df = Alpha_df.set_index("Timestamp")
Alpha_df = Alpha_df.sort_index()
```

```python
df = Alpha_df.resample('D').last().dropna()
```

```python
df['pnl'] = df[['portfolio_value']].diff()
df['pnl'].fillna(df['portfolio_value'], inplace=True)
```

```python
df[['portfolio_value', 'pnl']].head()
```

```python
import plotly.graph_objects as go
import pandas as pd

# Convert Pandas Series to List
dates = df.index.tolist()
pnl_values = df['pnl'].tolist()

# Create Bar Chart
fig = go.Figure()

fig.add_trace(go.Bar(
    x=dates,  
    y=pnl_values,  
    marker=dict(color=["green" if val >= 0 else "red" for val in pnl_values]),  # Color based on PnL
    name="Daily PnL"
))

# Customize Layout
fig.update_layout(
    title="Daily PnL",
    xaxis_title="Date",
    yaxis_title="PnL",
    xaxis=dict(tickformat="%Y-%m-%d"),  # Format x-axis as dates
    showlegend=False
)

# Show Plot
fig.show()
```

```python
df = df[['portfolio_value', 'pnl']]
```

```python
n_df = df[df['pnl'] < 0]

n_df[['pnl']]
```

```python
import plotly.graph_objects as go
import pandas as pd


# Convert PnL values to a list
pnl_values = n_df["pnl"].tolist()

# Create Histogram with Bin Size = 5
fig = go.Figure()

fig.add_trace(go.Histogram(
    x=pnl_values,  
    marker=dict(color="blue"),  # Set histogram bar color
    name="PnL Distribution",
    xbins=dict(  
        start=min(pnl_values),  # Start bin from minimum PnL value
        end=max(pnl_values),    # End bin at maximum PnL value
        size=500                 # Bin size of 5
    )
))

# Customize Layout
fig.update_layout(
    title="PnL Distribution Histogram (Bin Size = 5)",
    xaxis_title="PnL",
    yaxis_title="Frequency",
    bargap=0.1,  # Adjust gap between bars
    showlegend=False
)

# Show Plot
fig.show()
```
