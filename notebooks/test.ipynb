```python
import pandas as pd
```

```python
pv_old = pd.Series(range(0,10,1))
pv_new = pd.Series([4,4,4,6,2,8,4,6,8,0])

pv_old.plot()
pv_new.plot()
```

```python
daily_pnl_old = pv_old.diff()
daily_pnl_new = pv_new.diff()

daily_pnl_old.plot()
daily_pnl_new.plot()
```

```python
gap_in_daily_pnls = daily_pnl_new - daily_pnl_old
gap_in_daily_pnls.plot()
```

```python
import talib
import pandas as pd

# Sample Data
data = {
    'close': [100, 102, 104, 103, 105, 107, 110, 108, 109, 111]
}
df = pd.DataFrame(data)

# Define period
period = 3

# Compute SMA
df['SMA'] = talib.SMA(df['close'], timeperiod=period)

# Compute EMA
df['EMA'] = talib.EMA(df['close'], timeperiod=period)

# Display Result
print(df)
```
