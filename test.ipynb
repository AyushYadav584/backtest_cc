```python
import importlib.util
import numpy as np

def load_variables(file_path):
    """Load variables from a Python file dynamically."""
    spec = importlib.util.spec_from_file_location("module_name", file_path)
    module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(module)
    return {k: v for k, v in vars(module).items() if not k.startswith("__")}

def compare_files(file1, file2):
    """Compare key-value pairs between two Python files."""
    data1 = load_variables(file1)
    data2 = load_variables(file2)
    
    all_keys = set(data1.keys()).union(set(data2.keys()))
    
    for key in all_keys:
        val1 = data1.get(key, "Not Found")
        val2 = data2.get(key, "Not Found")
        
        # Handle NumPy arrays separately
        if isinstance(val1, np.ndarray) or isinstance(val2, np.ndarray):
            if not np.array_equal(val1, val2):
                print(f"Difference in {key}: File1={val1}, File2={val2}\n")
        else:
            if val1 != val2:
                print(f"Difference in {key}: File1={val1}, File2={val2}\n")

# Example usage
# compare_files("file1.py", "file2.py")
```

```python
def get_transaction_fee_blotter(folder_path):
    try:
        # print(folder_path)
        Blotter_df_list = glob.glob(folder_path + '/blotter/*.csv')
        Blotter_bt_df_list = [pd.read_csv(i) for i in Blotter_df_list]
        Blotter_df = pd.concat(Blotter_bt_df_list, ignore_index=True)

        Blotter_df = Blotter_df[Blotter_df['trade_sub_type'] == 'trade']

    except Exception as e:
        print(f"Error: {folder_path} {e}")


import pandas as pd

def get_trade_statistics(df):
    """
    Returns daily statistics: 
    - Count of rows where trade_sub_type == 'trade'
    - Count of trades where position < 0
    - Count of trades where position > 0
    - Aggregated max, mean, and min for all metrics

    Parameters:
    df (pd.DataFrame): DataFrame with 'trade_sub_type' and 'position' columns, indexed by timestamp.

    Returns:
    pd.DataFrame: DataFrame with trade statistics per date.
    pd.DataFrame: DataFrame with max, mean, and min for trade statistics.
    """
    # Ensure index is a DateTimeIndex
    Blotter_df_list = glob.glob(folder_path + '/blotter/*.csv')
    Blotter_bt_df_list = [pd.read_csv(i) for i in Blotter_df_list]
    df = pd.concat(Blotter_bt_df_list, ignore_index=True)

    df.set_index('timestamp', inplace=True)

    df.index = pd.to_datetime(df.index)

    # Extract date from index
    df['date'] = df.index.date

    # Filter only rows where trade_sub_type == 'trade'
    trade_df = df[df['trade_sub_type'] == 'trade']

    # Group by date and calculate required metrics
    trade_stats = trade_df.groupby('date').agg(
        trade_count=('trade_sub_type', 'count'),
        negative_trades=('position', lambda x: (x < 0).sum()),
        positive_trades=('position', lambda x: (x > 0).sum())
    )

    # Calculate overall max, mean, and min for each metric
    summary_stats = trade_stats.agg(['max', 'mean', 'min'])

    return trade_stats, summary_stats

# Example usage
data = {
    'trade_sub_type': ['trade', 'trade', 'other', 'trade', 'trade', 'trade', 'trade', 'other', 'trade'],
    'position': [100, -50, 200, -30, 70, -10, -20, 40, 60]
}
timestamps = [
    '2024-02-01 10:00:00', '2024-02-01 10:05:00', '2024-02-01 10:10:00',
    '2024-02-02 11:00:00', '2024-02-02 11:05:00', '2024-02-02 11:10:00',
    '2024-02-03 12:00:00', '2024-02-03 12:05:00', '2024-02-03 12:10:00'
]
df = pd.DataFrame(data, index=pd.to_datetime(timestamps))

trade_stats, summary_stats = get_trade_statistics(df)

print("Trade Statistics Per Date:\n", trade_stats)
print("\nSummary Statistics (Max, Mean, Min):\n", summary_stats)
```

```python
import pandas as pd

# Sample DataFrame
# df = pd.read_csv("/home/cloudcraftz/Office_Projects/10_Fintech/backtest_2.0_dev/HFT-Options-EIS-Global/tradelib/outputs/NF_SMA_CROSS_OVER_MFT_MT/2024_NIFTY_SMA_STRADDLE_STRATEGY_fast_25_slow_40_tol_0_mt_1000/blotter/blotter_20240101092000_20240101153000.csv")

def trade_duration(df):

    df['timestamp'] = pd.to_datetime(df['timestamp'])
    # Extract the trade date for grouping
    df['date'] = df['timestamp'].dt.date

    total_number_of_trading_days = df['date'].nunique()
    # print(f"{total_number_of_trading_days=}")

    # Filtering trades and unwinds separately
    trades = df[df['trade_sub_type'] == 'trade'][['date', 'timestamp', 'trade_sub_type']].drop_duplicates().sort_values('timestamp').reset_index(drop=True)
    unwinds = df[df['trade_sub_type'] == 'unwind'][['date', 'timestamp', 'trade_sub_type']].drop_duplicates().sort_values('timestamp').reset_index(drop=True)

    # # # Calculating duration
    trades['duration_minutes'] = (unwinds['timestamp'] - trades['timestamp']).dt.total_seconds()//60

    # Categorizing trades based on duration
    trades['lt_2min'] = trades['duration_minutes'] < 2
    trades['lt_5min'] = trades['duration_minutes'] < 5

    # Grouping by date
    daily_counts = trades.groupby('date').agg({'lt_2min': 'sum', 'lt_5min': 'sum'})

    # Compute daily averages
    avg_lt_2min = daily_counts['lt_2min'].mean()
    avg_lt_5min = daily_counts['lt_5min'].mean()

    # Count number of days with at least one trade in each category
    days_with_lt_2min = (daily_counts['lt_2min'] > 0).sum()
    days_with_lt_5min = (daily_counts['lt_5min'] > 0).sum()

    return round(avg_lt_2min,2), round(avg_lt_5min, 2), days_with_lt_2min, days_with_lt_5min, round(days_with_lt_2min/total_number_of_trading_days*100, 2), round(days_with_lt_5min/total_number_of_trading_days*100, 2)

    # # Print results
    # print(f"Average daily trades < 2 mins: {avg_lt_2min:.2f}")
    # print(f"Average daily trades < 5 mins: {avg_lt_5min:.2f}")
    # print(f"Days with at least one trade < 2 mins: {days_with_lt_2min}")
    # print(f"Days with at least one trade < 5 mins: {days_with_lt_5min}")
    # print(f"Days with at least one trade < 2 mins: {round(days_with_lt_2min/total_number_of_trading_days*100, 2)}")
    # print(f"Days with at least one trade < 5 mins: {round(days_with_lt_5min/total_number_of_trading_days*100, 2)}")

trade_duration(df)
```

```python
import numpy as np
import pandas as pd
from scipy import stats

class MovingAverageSlopeStrategy:
    def __init__(self, window=200):
        self.window = window

    def _linear_regression_slope(self, y):
        x = np.arange(len(y))
        slope, _, _, _, _ = stats.linregress(x, y)
        return slope

    def calculate_trend_slope(self, series):
        return series.rolling(window=self.window).apply(self._linear_regression_slope, raw=True)

    def generate_signals(self, stock_data):
        stock_data['Slope'] = self.calculate_trend_slope(stock_data['Close'])
        stock_data['Signal'] = 0
        buy_signal_triggered = False

        for i in range(1, len(stock_data)):
            if not buy_signal_triggered and stock_data['Slope'].iloc[i] > 0:
                stock_data.loc[stock_data.index[i], 'Signal'] = 1  # Buy signal
                buy_signal_triggered = True
            elif buy_signal_triggered and stock_data['Slope'].iloc[i] < 0:
                stock_data.loc[stock_data.index[i], 'Signal'] = -1  # Sell signal
                buy_signal_triggered = False
        
        return stock_data

# Example usage:
# df = pd.read_csv('your_data.csv')  # Ensure 'Close' column exists
# strategy = MovingAverageSlopeStrategy(window=200)
# signals = strategy.generate_signals(df)
```
