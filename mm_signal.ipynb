```python
import numpy as np
import pandas as pd
from scipy import stats
from collections import deque

class MovingAverageSlopeSignal:
    def __init__(self, window=200):
        self.window = window
        self.values = deque(maxlen=window)
        self.buy_signal_triggered = False

    def _linear_regression_slope(self, y):
        x = np.arange(len(y))
        slope, _, _, _, _ = stats.linregress(x, y)
        return slope

    def calculate_trend_slope(self):
        if len(self.values) < self.window:
            return None
        series = pd.Series(self.values)
        return self._linear_regression_slope(series)
    
    def add(self, value):
        self.values.append(value)
    
    def get_signal(self):
        slope = self.calculate_trend_slope()
        if slope is None:
            return 0  # No signal until enough data is available
        
        if not self.buy_signal_triggered and slope > 0:
            self.buy_signal_triggered = True
            return 1  # Buy signal
        elif self.buy_signal_triggered and slope < 0:
            self.buy_signal_triggered = False
            return -1  # Sell signal
        return 0  # No signal

# Example usage:
if __name__ == "__main__":
    strategy = MovingAverageSlopeSignal(window=5)
    data = [100, 102, 101, 105, 107, 108]
    
    for value in data:
        strategy.add(value)
        print(f"Value: {value}, Signal: {strategy.get_signal()}")
```
